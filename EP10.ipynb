```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('auto_mpg.csv')

sns.boxplot(df['horsepower'])

plt.show()
```

```python
df['horsepower']=pd.to_numeric(df['horsepower'],errors='coerce')
Q1=df['horsepower'].quantile(0.25)
Q3=df['horsepower'].quantile(0.75)
iqr=Q3-Q1
lower_limit=Q1-(1.5*iqr)
upper_limit=Q3+(1.5*iqr)
df[(df['horsepower']>upper_limit)|(df['horsepower']<lower_limit)]
lower_cap=df['horsepower'].quantile(0.1)
upper_cap=df['horsepower'].quantile(0.9)
df['horsepower']=np.where(df['horsepower']>upper_limit,upper_cap,np.where(df['horsepower']<lower_limit,lower_cap,df['horsepower']))
sns.boxplot(df['horsepower'])
```

```python
print("before replacing outliers")
sns.boxplot(df['acceleration'])
plt.show()

q1 = df['acceleration'].quantile(0.25)
q3 = df['acceleration'].quantile(0.75)

iqr = q3 - q1

lower_limit = q1 - (1.5 * iqr)
upper_limit = q3 + (1.5 * iqr)

df[(df['acceleration'] > upper_limit) |
   (df['acceleration'] < lower_limit)]

upper_cap = df['acceleration'].quantile(0.9)
lower_cap = df['acceleration'].quantile(0.1)

df['acceleration'] = np.where(
    df['acceleration'] > upper_limit,
    upper_cap,
    np.where(
        df['acceleration'] < lower_limit,
        lower_cap,
        df['acceleration']
    )
)

print("after replacing outliers")
sns.boxplot(df['acceleration'])
```
